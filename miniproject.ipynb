```python
import numpy as np
import matplotlib.pyplot as plt
from sklearn.linear_model import LogisticRegression

X = np.array([[85], [89], [90], [95], [100], [110], [120], [130], [140], [150]])
y = np.array(['no', 'no', 'no', 'no', 'no', 'yes', 'yes', 'yes', 'yes', 'yes'])

model = LogisticRegression()
model.fit(X, y)
glucose = int(input("Enter your glucose level: "))

prediction = model.predict([[glucose]])[0]
prob = model.predict_proba([[glucose]])[0][1]

print("Prediction:", prediction)
if prediction == 'yes':
    print(" You are likely diabetic.")
else:
    print(" You are likely not diabetic.")

test_range = np.linspace(70, 180).reshape(-1, 1)
prob_curve = model.predict_proba(test_range)[:, 1]

plt.scatter(X, [1 if i == 'yes' else 0 for i in y], color='red', label='Training Data')
plt.plot(test_range, prob_curve, color='blue', label='Probability Curve')
plt.axvline(glucose, color='green', linestyle='--', label=f'You: {glucose}')
plt.xlabel("Glucose Level")
plt.ylabel("Probability of Diabetes")
plt.title("Logistic Regression: Diabetes Prediction")
plt.legend()
plt.grid(True)
plt.show()
```
